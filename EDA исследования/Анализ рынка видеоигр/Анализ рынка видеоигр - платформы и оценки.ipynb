Большое количество пропусков в Critic_Score(51.39%), User_Score (54.66%), Rating (40.52%). Возможно, оценки и рейтинги не были присвоены для многих игр.

Решаем: 
   - Пропуски в оценках и рейтингах оставляем как NaN, так как их удаление сильно сократит выборку.
   - Пропуски в продажах (eu_sales, jp_sales) заполним медианой по платформе и году.

- Обработайте пропущенные значения. Для каждого случая вы можете выбрать оптимальный, на ваш взгляд, вариант: заменить на определённое значение, оставить как есть или удалить.
- Если вы решите заменить пропуски на значение-индикатор, то убедитесь, что предложенное значение не может быть использовано в данных.
- Если вы нашли пропуски в данных с количеством проданных копий игры в том или ином регионе, их можно заменить на среднее значение в зависимости от названия платформы и года выхода игры.

```python
# Удаляем пропуски в name и genre (их мало)
initial_rows = len(df)
df = df.dropna(subset=['name', 'genre'])
print(f"Удалено строк с пропусками в name/genre: {initial_rows - len(df)}")
```

```python
# Заполняем пропуски в продажах медианой по платформе и году выпуска
sales_columns = ['na_sales', 'eu_sales', 'jp_sales', 'other_sales']
for col in sales_columns:
    nulls_before = df[col].isnull().sum()
    #Медиана по платформе и году выпуска
    df[col] = df.groupby(['platform', 'year_of_release'])[col].transform(
        lambda x: x.fillna(x.median())
    )
    #Медиана по платформе
    if df[col].isnull().sum() > 0:
        df[col] = df.groupby('platform')[col].transform(
            lambda x: x.fillna(x.median())
        )
    #Общая медиана
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
    
    nulls_after = df[col].isnull().sum()
    print(f"  {col}: было {nulls_before} пропусков, стало {nulls_after}")
```

## 4. Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
def user_category(score):
    if pd.isnull(score):
        return 'не оценено'
    elif 8 <= score <= 10:
        return 'высокая'
    elif 3 <= score < 8:
        return 'средняя'
    elif 0 <= score < 3:
        return 'низкая'
    else:
        return 'неизвестно'
```

```python
df_actual['user_category'] = df_actual['user_score'].apply(user_category)
```

- Разделите все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
def critic_category(score):
    if pd.isnull(score):
        return 'не оценено'
    elif 80 <= score <= 100:
        return 'высокая'
    elif 30 <= score < 80:
        return 'средняя'
    elif 0 <= score < 30:
        return 'низкая'
    else:
        return 'неизвестно'
```

```python
df_actual['critic_category'] = df_actual['critic_score'].apply(critic_category)
```

- После категоризации данных проверьте результат: сгруппируйте данные по выделенным категориям и посчитайте количество игр в каждой категории.

```python
print("Категоризация по оценкам пользователей:")
user_cat_counts = df_actual['user_category'].value_counts()
for category, count in user_cat_counts.items():
    percentage = (count / len(df_actual)) * 100
    print(f"{category:<15} {count:>6} игр ({percentage:>5.1f}%)")
```

```python
print("\nКатегоризация по оценкам критиков:")
critic_cat_counts = df_actual['critic_category'].value_counts()
for category, count in critic_cat_counts.items():
    percentage = (count / len(df_actual)) * 100
    print(f"{category:<15} {count:>6} игр ({percentage:>5.1f}%)")
```

- Выделите топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
top_platforms = df_actual['platform'].value_counts().head(7)
platform_stats = pd.DataFrame({
    'Количество игр': top_platforms.values,
    'Доля от всех игр (%)': (top_platforms.values / len(df_actual) * 100).round(1)
}, index=top_platforms.index)
```
